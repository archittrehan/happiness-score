```python
import numpy as np
import pandas as pd

df_2015 = pd.read_csv("datasets/2015.csv")
df_2016 = pd.read_csv("datasets/2016.csv")
df_2017 = pd.read_csv("datasets/2017.csv")
df_2018 = pd.read_csv("datasets/2018.csv")
df_2019 = pd.read_csv("datasets/2019.csv")
    
```

```python
#renaming various columns in the datasets
print(df_2015.columns)
print(df_2016.columns)
print(df_2017.columns)
print(df_2018.columns)
print(df_2019.columns)
df_2015 = df_2015.rename(columns={
    "Happiness Score": "Score",
    "Economy (GDP per Capita)": "GDP per capita",
    "Health (Life Expectancy)" : "Healthy life expectancy",
    "Trust (Government Corruption)" : "Perceptions of corruption"
})
df_2016 = df_2016.rename(columns={
    "Happiness Score": "Score",
    "Economy (GDP per Capita)": "GDP per capita",
    "Health (Life Expectancy)" : "Healthy life expectancy",
    "Trust (Government Corruption)" : "Perceptions of corruption"
})
df_2017 = df_2017.rename(columns={
    "Happiness.Score": "Score",
    "Economy..GDP.per.Capita.": "GDP per capita",
    "Health..Life.Expectancy." : "Healthy life expectancy",
    "Trust..Government.Corruption." : "Perceptions of corruption"
})
df_2018 = df_2018.rename(columns={
    "Freedom to make life choices":"Freedom",
    "Country or region" : "Country"
})
df_2019 = df_2019.rename(columns={
    "Freedom to make life choices":"Freedom",
    "Country or region" : "Country"
})
```

```python
#combining the different data-sets 
df=pd.concat([df_2015,df_2016,df_2017,df_2018,df_2019])
df =df.sort_values(by = ['Country'])


df=df.reset_index()
df.head(25)
```

```python
corr_matrix = df.corr(numeric_only=True)
print(corr_matrix)
#print(df.columns)
#strongest relations with score (possible input features, ex-> happiness rank might not be a possible input feature):- gdp per capita,Healthy life expectancy . Also with this we are automatically removing columns with over 60 percent of NaN values.
df = df.drop(columns=['Happiness Rank','Overall rank','Social support','Happiness.Rank','Perceptions of corruption','Generosity','Freedom','Family','Region','Standard Error','Dystopia Residual','Lower Confidence Interval','Upper Confidence Interval','Whisker.high','Whisker.low','Dystopia.Residual'])
#dropping the outliers as they are unique cases skewing the data and not something that would affect the learning model as a whole as they are not some special minority case
cols = ["GDP per capita",  "Healthy life expectancy"]

df_clean = df.copy()

for col in cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    
    df_clean = df_clean[(df_clean[col] >= lower) & (df_clean[col] <= upper)]
```

```python
print(df.columns)
```

```python
#the data is ready. building the model
features = ["GDP per capita","Healthy life expectancy"]
target= "Score"
df=df_clean[features + [target]].dropna()
```

```python
#splitting into train and test
indices = np.arange(len(df))
np.random.shuffle(indices)

split = int(0.8 * len(df))

train_idx = indices[:split]
test_idx = indices[split:]

train = df.iloc[train_idx]
test = df.iloc[test_idx]
```

```python
X_train = train[features].values
X_test = test[features].values

mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std
```

```python
X_train = np.c_[np.ones(X_train.shape[0]), X_train]
X_test = np.c_[np.ones(X_test.shape[0]), X_test]
```

```python
beta = np.linalg.lstsq(X_train, train[target].values, rcond=None)[0]
```

```python
y_test = test[target].values
y_pred = X_test @ beta
```

```python
mse = np.mean((y_test - y_pred) ** 2)
print("MSE:", mse)
```

```python
ss_res = np.sum((y_test - y_pred) ** 2)
ss_tot = np.sum((y_test - y_test.mean()) ** 2)

r2 = 1 - (ss_res / ss_tot)
print("R²:", r2)
```

```python
plt.figure(figsize=(6,6))
plt.scatter(y_test, y_pred)

# perfect line
plt.plot([y_test.min(), y_test.max()],
         [y_test.min(), y_test.max()],
         'r--')

plt.xlabel("Actual")
plt.ylabel("Predicted")

plt.show()
```

```python
#using scikit learn
x_train = train[features].values
x_test = test[features].values

Y_train = train[target].values
Y_test = test[target].values
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)  
x_test = scaler.transform(x_test)        
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(x_train, Y_train)
```

```python
Y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(Y_test, Y_pred)
r2 = r2_score(Y_test, Y_pred)

print("MSE:", mse)
print("R²:", r2)
```
